```python
import pandas as pd
import numpy as np
from pathlib import Path

prices = pd.read_csv("../data/processed/prices.csv", index_col="date", parse_dates=True)
prices.shape
```

```python
log_returns = np.log(prices / prices.shift(1))
log_returns = log_returns.iloc[1:]  # drop the first row, which is NaN (nothing to compare it to)
log_returns.head()
```

```python
def avg_autocorr(returns, lag):
    """Average autocorrelation across all coins at a given lag."""
    return returns.apply(lambda col: col.autocorr(lag=lag)).mean()

horizons = {"1-day": 1, "1-week": 7, "1-month": 30, "3-month": 90}
results = {name: avg_autocorr(log_returns, lag) for name, lag in horizons.items()}
pd.Series(results)
```

```python
N = len(log_returns)
significance_threshold = 1.96 / np.sqrt(N)
print(f"Rough significance threshold at this sample size: ±{significance_threshold:.4f}")
```
